```python
import re
import nltk
import pandas as pd
import matplotlib.pyplot as plt
from nltk.corpus import movie_reviews, stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.probability import FreqDist
```

```python
resources = ["movie_reviews", "stopwords", "punkt", "punkt_tab", "wordnet", "omw-1.4", "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]
for resource in resources:
    nltk.download(resource, quiet=True)
```

```python
data = [
    (" ".join(movie_reviews.words(fileid)), category)
    for category in movie_reviews.categories()
    for fileid in movie_reviews.fileids(category)
]
df = pd.DataFrame(data, columns=["text", "label"])
df.head()
```

```python
sample = "<p>I LOVED this movie!!! Visit https://example.com 😊 It was REALLY good.</p>"
clean = re.sub(r"<[^>]+>", " ", sample)
clean = re.sub(r"https?://\S+|www\.\S+", " ", clean)
clean = re.sub(r"[^A-Za-z\s]", " ", clean)
clean = re.sub(r"\s+", " ", clean).strip()
print(clean)
```

```python
normalized = clean.lower()
normalized
```

```python
text = "NLP is interesting. Text processing is the first step."
print(sent_tokenize(text))
print(word_tokenize(text))
```

```python
stop_words = set(stopwords.words("english")) - {"not", "no", "nor"}
tokens = word_tokenize(normalized)
filtered = [word for word in tokens if word not in stop_words]
filtered
```

```python
stemmer = PorterStemmer()
words = ["playing", "played", "plays", "studies", "better"]
[(word, stemmer.stem(word)) for word in words]
```

```python
lemmatizer = WordNetLemmatizer()
words = ["cars", "running", "better", "studies"]
[(word, lemmatizer.lemmatize(word)) for word in words]
```

POS Taging

```python
from nltk import pos_tag

def wordnet_pos(tag):
    return {"J": "a", "V": "v", "N": "n", "R": "r"}.get(tag[0], "n")

sentence = "The cars were running and the actors were better"
tagged = pos_tag(word_tokenize(sentence))
[(word, lemmatizer.lemmatize(word, wordnet_pos(tag))) for word, tag in tagged]
```
